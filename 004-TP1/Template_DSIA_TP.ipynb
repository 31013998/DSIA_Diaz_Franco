## 📝 Descripción del Dataset

Para este trabajo se utilizará un conjunto de datos de exoplanetas confirmados por la NASA hasta el año 2021\.

Un exoplaneta es un planeta que orbita una estrella distinta al Sol\. El catálogo reúne información física y orbital de miles de exoplanetas descubiertos mediante diferentes técnicas de observación astronómica\.

El dataset contiene variables relacionadas con:

Características del planeta \(masa, radio, densidad, período orbital, etc\.\)\.
Características de la estrella anfitriona \(masa, radio, temperatura efectiva, metalicidad, etc\.\)\.
Método de descubrimiento\.
Distancia al sistema planetario\.
El objetivo del trabajo es realizar un análisis exploratorio de los datos, identificar patrones relevantes y formular hipótesis que permitan comprender mejor las relaciones existentes entre las variables astronómicas\.
