Ejercicio 1 – Variables y operaciones

```
nombre = input("Ingresa tu nombre: ")
edad = int(input("Ingresa tu edad: "))
ciudad = input("Ingresa tu ciudad: ")

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")

while True:
    try:
        a = float(input("Ingresa el número a: "))
        break
    except ValueError:
        print("Solo se pueden ingresar números.")


while True:
    try:
        b = float(input("Ingresa el número b: "))
        break
    except ValueError:
        print("Solo se pueden ingresar números.")


suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)
```

Ejercicio 2 – Condicionales

```
while True:
    try:
        edad = int(input("Ingresa tu edad: "))
        break
    except ValueError:
        print("Solo se pueden ingresar números.")

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

Ejercicio 3 – Listas y bucles

```
numeros = [2, 8, 19, 26, 33]

print("Números de la lista:")

for numero in numeros:
    print(numero)

suma = sum(numeros)
promedio = suma / len(numeros)

print("Suma total:", suma)
print("Promedio:", promedio)
```
