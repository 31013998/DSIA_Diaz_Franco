```

data = {
    'estudiante': ['Constanza', 'Jorge', 'Federico', 'Catalina', 'Franco'],
    'nota': [8, 6, 9, 5, 7]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)
print()

promedio = df['nota'].mean()
print(f"Promedio de notas: {promedio}")

nota_max = df['nota'].max()
print(f"Nota máxima: {nota_max}")

nota_min = df['nota'].min()
print(f"Nota mínima: {nota_min}")
```
